## Make subsets of TF-ACRs only sensitive to a given TF, for use in the NFE2 analyses

```python
import pandas as pd
import os
```

```python
eryth_dir = "/mnt/md0/varshini/Analysis/Blood/complete_analyses/track_defs/erythroid_tracks/classifications"
```

```python
eryth_tf = pd.read_csv(os.path.join(eryth_dir, "erythroid_tf_acrs_all_enhancernotpromoternotctcf.bed"), sep="\t",
                      names=['chr','start','end','pval','perturbation'])
tf = pd.read_csv(os.path.join(eryth_dir, "tf_acrs_sorted_enhancernotpromoternotctcf.bed"), sep="\t",
                names=['chr','start','end','pval','perturbation'])
```

```python
tf_nfe2 = tf[tf['perturbation']=='NFE2']
eryth_nfe2 = eryth_tf[eryth_tf['perturbation']=='NFE2']
```

```python
# now nfe2-only
tf_nfe2_only = tf.drop_duplicates(subset=['chr', 'start', 'end'], keep=False)
tf_nfe2_only = tf_nfe2_only[tf_nfe2_only['perturbation']=='NFE2']

eryth_nfe2_only = eryth_tf.drop_duplicates(subset=['chr', 'start', 'end'], keep=False)
eryth_nfe2_only = eryth_nfe2_only[eryth_nfe2_only['perturbation']=='NFE2']
```

```python
eryth_nfe2_only
```

```python
eryth_nfe2_only.to_csv("/mnt/md0/varshini/Analysis/Blood/complete_analyses/track_defs/erythroid_tracks/classifications/nfe2_only_eryth.bed",
                      index=False,
                      header=False,
                      sep="\t")

tf_nfe2_only.to_csv("/mnt/md0/varshini/Analysis/Blood/complete_analyses/track_defs/erythroid_tracks/classifications/nfe2_only_tf.bed",
                      index=False,
                      header=False,
                      sep="\t")
```

```python
# now something else-only
tf_ctrl_only = tf.drop_duplicates(subset=['chr', 'start', 'end'], keep=False)
tf_ctrl_only = tf_ctrl_only[tf_ctrl_only['perturbation']=='GFI1B']

eryth_ctrl_only = eryth_tf.drop_duplicates(subset=['chr', 'start', 'end'], keep=False)
eryth_ctrl_only = eryth_ctrl_only[eryth_ctrl_only['perturbation']=='GFI1B']


eryth_ctrl_only.to_csv("/mnt/md0/varshini/Analysis/Blood/complete_analyses/track_defs/erythroid_tracks/classifications/gfi1b_only_eryth.bed",
                      index=False,
                      header=False,
                      sep="\t")



tf_ctrl_only.to_csv("/mnt/md0/varshini/Analysis/Blood/complete_analyses/track_defs/erythroid_tracks/classifications/gfi1b_only_tf.bed",
                      index=False,
                      header=False,
                      sep="\t")
```
