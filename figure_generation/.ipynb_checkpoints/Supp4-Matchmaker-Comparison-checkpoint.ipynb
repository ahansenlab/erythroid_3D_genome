## triangle scores

```python
paul_tol_pal = ["#CC3311", "#0077BB", "#009988", "#33BBEE", "#BBBBBB", "#EE3377", "#EE7733"]
```

```python
paul_tol_pal = ['#33BBEE', "#0077BB",  "#BBBBBB","#EE3377", "#CC3311", "#EE7733"]

labels_plot = ['eryTF-CREs',
               'ATAC-noCTCF',
               'random',
               'prom_accesible',
               'CTCF',
                'random'
              ]

for i in [1, 2, 3]:

    score_col_phase = f"triangle_score_P{i}"
    
    df_tidy = pd.DataFrame(columns=['variable','value'])
    for label in labels_plot:

        curr_df = pd.DataFrame(columns=['variable','value'])
        curr_df['value'] = match_scores_all[label][score_col_phase].dropna()
        curr_df['variable'] = label
        
        df_tidy = pd.concat([df_tidy, curr_df], axis=0)
    print(df_tidy.groupby('variable').agg('mean'))
    #plt.figure(figsize=(5,6))
    #sns.violinplot(df_tidy, x='variable', y='value',palette=paul_tol_pal, hue='variable')
    #plt.show()
    
    plt.figure(figsize=(6,6))
    sns.boxplot(df_tidy, x='variable', y='value', hue='variable', palette=paul_tol_pal,
                showfliers=False)
    plt.xticks(rotation=45, ha='right') 
    plt.ylim([0, 2.25])
    plt.savefig(os.path.join(savedir, f"comparison_boxplot_P{i}.svg"))
    plt.show()
    # get the score col from each nametag into a df
    
    p_values = []
    comparison_names = []
    
    groups = [group_df['value'].values for name, group_df in df_tidy.groupby('variable')]
    names = [name for name, group_df in df_tidy.groupby('variable')]
    pairs = list([i for i in itertools.combinations(range(len(groups)), r=2)])

    for pair in pairs:
            g1, g2 = pair
            stat, p = stats.mannwhitneyu(groups[g1], groups[g2])
            p_values.append(p)
            comparison_names.append(f"{names[g1]} vs {names[g2]}")

    reject, p_corrected, alpha_sidak, alpha_bonf = multitest.multipletests(
            p_values, alpha=0.05, method='fdr_bh'
        )

    print("\nPost-hoc Mann-Whitney U (BH Corrected):")
    for name, p, corr_p, rej in zip(comparison_names, p_values, p_corrected, reject):
            print(f"{name}: raw p = {p}, corrected p = {corr_p}, significant = {rej}")
```
