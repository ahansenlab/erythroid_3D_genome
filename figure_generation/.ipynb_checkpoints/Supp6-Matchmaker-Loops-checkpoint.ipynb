```python
def explode_df(df, p):
    df = df.copy()
    exploded = df.explode([f'crosser_loop_scores_{p}', f'crosser_loop_coords_{p}', f'crosser_loop_classes_{p}']).dropna(subset=f'crosser_loop_coords_{p}')

    exploded['coord1'] = exploded[f'crosser_loop_coords_{p}'].apply(lambda x: x[0])
    exploded['coord2'] = exploded[f'crosser_loop_coords_{p}'].apply(lambda x: x[1])
    exploded['distance'] = exploded['coord2'] - exploded['coord1']
    exploded.rename({'chr':'chrom1'}, inplace=True, axis=1)
    
    return exploded

def explode_ss(df, p):
    df = df.copy()
    exploded = df.explode([f'ss_loop_scores_{p}', f'ss_loop_coords_{p}', f'ss_loop_classes_{p}']).dropna(subset=f'ss_loop_coords_{p}')

    exploded['coord1'] = exploded[f'ss_loop_coords_{p}'].apply(lambda x: x[0])
    exploded['coord2'] = exploded[f'ss_loop_coords_{p}'].apply(lambda x: x[1])
    exploded['distance'] = exploded['coord2'] - exploded['coord1']
    exploded.rename({'chr':'chrom1'}, inplace=True, axis=1)
    
    return exploded
```

```python
def get_corrs(mm_loops, p, lt='crosser', ylim_able=None, ylim_oe=None, writepath=None, agg_fun='mean'):
    
    # AbLE
    plt.scatter(mm_loops.groupby(['chrom1','coord']).agg(var = (f'triangle_score_{p}', 'mean'))['var'].values, 
                mm_loops.groupby(['chrom1','coord']).agg(var = (f'strength_{p}', agg_fun))['var'].values, alpha=0.5)
    if ylim_able:
        plt.ylim(ylim_able)
    if writepath:
        plt.savefig(os.path.join(writepath, f"able_corr_{lt}_{p},.svg"))   
    plt.show()
    
    # O/E
    plt.scatter(mm_loops.groupby(['chrom1','coord']).agg(var = (f'triangle_score_{p}', 'mean'))['var'].values, 
                mm_loops.groupby(['chrom1','coord']).agg(var = (f'{lt}_loop_scores_{p}', agg_fun))['var'].values)
    
    if ylim_oe:
        plt.ylim(ylim_oe)
    if writepath:
        plt.savefig(os.path.join(writepath, f"oe_corr_{lt}_{p}.svg"))
    plt.show()
    
    # O/E regression coeffs
    clean_x = mm_loops.dropna(subset=[f'triangle_score_{p}', f'{lt}_loop_scores_{p}'])[f'triangle_score_{p}'].values.reshape(-1, 1)
    clean_y = mm_loops.dropna(subset=[f'triangle_score_{p}', f'{lt}_loop_scores_{p}'])[f'{lt}_loop_scores_{p}'].values.reshape(-1, 1)
    reg = LinearRegression().fit(clean_x, clean_y)
    interc = reg.intercept_
    coef = reg.coef_
    pred_y = reg.predict(clean_x)
    r2_oe = reg.score(clean_x, clean_y)

    # able regression coeffs
    clean_x = mm_loops.dropna(subset=[f'triangle_score_{p}', f'strength_{p}'])[f'triangle_score_{p}'].values.reshape(-1, 1)
    clean_y = mm_loops.dropna(subset=[f'triangle_score_{p}', f'strength_{p}'])[f'strength_{p}'].values.reshape(-1, 1)
    reg = LinearRegression().fit(clean_x, clean_y)
    interc = reg.intercept_
    coef = reg.coef_
    pred_y = reg.predict(clean_x)
    r2_able = reg.score(clean_x, clean_y)

    return r2_able, r2_oe


def plots_crosser_ss(mm_loops, exploded_merged_control, p, lt='crosser', writepath=None, tag='mm'):
 
    sns.histplot(mm_loops[f'{lt}_loop_scores_{p}'].dropna(), bins=50, label=f'{p} MM',
            stat='density', element='step', fill=False)
    sns.histplot(exploded_merged_control[f'{lt}_loop_scores_{p}'].dropna(), bins=50, label=f'{p} {control_nametag}',
            stat='density', element='step', fill=False)
    plt.legend()
    plt.show()

    sns.histplot(mm_loops[f'strength_{p}'].dropna(), bins=50, label=f'{p} MM',
                stat='density', element='step', fill=False)
    sns.histplot(exploded_merged_control[f'strength_{p}'].dropna(), bins=50, label=f'{p} {control_nametag}',
                stat='density', element='step', fill=False)
    plt.legend()
    plt.show()

    # violin plot of loop scores O/E
    tidy_mm = pd.DataFrame()
    tidy_mm['value'] = mm_loops[f'{lt}_loop_scores_{p}'].dropna()
    tidy_mm['variable'] = 'MM'
    tidy_c = pd.DataFrame()
    tidy_c['value'] = exploded_merged_control[f'{lt}_loop_scores_{p}'].dropna()
    tidy_c['variable'] = control_nametag
    plt.figure(figsize=(4,4))
    tidy_plot = pd.concat([tidy_mm, tidy_c])
    
    print(tidy_plot)
    sns.violinplot(tidy_plot, x='variable', y='value', hue='variable', palette=pal[lt])
    plt.ylim(-0.1, 20)
    print(tidy_plot.groupby('variable')['value'].mean())
    if writepath:
        plt.savefig(os.path.join(writepath, f"oe_violin_{lt}_{p}_{tag}.svg"))
    plt.show()
    print(mannwhitneyu(mm_loops[f'{lt}_loop_scores_{p}'].fillna(value=np.nan).dropna().values, 
                       exploded_merged_control[f'{lt}_loop_scores_{p}'].fillna(value=np.nan).dropna().values,
         alternative='greater'))
    
    # violin plot of loop strengths AbLE
    tidy_mm = pd.DataFrame()
    tidy_mm['value'] = mm_loops[f'strength_{p}'].dropna()
    tidy_mm['variable'] = 'MM'
    tidy_c = pd.DataFrame()
    tidy_c['value'] = exploded_merged_control[f'strength_{p}'].dropna()
    tidy_c['variable'] = control_nametag
    plt.figure(figsize=(4,4))
    tidy_plot = pd.concat([tidy_mm, tidy_c])
    sns.violinplot(tidy_plot, x='variable', y='value', hue='variable', palette=pal[lt])
    plt.ylim(-0.1, 0.5)
    print(tidy_plot.groupby('variable')['value'].mean())
    if writepath:
        plt.savefig(os.path.join(writepath, f"able_violin_{lt}_{p}_{tag}.svg"))
    plt.show()
    print(mannwhitneyu(mm_loops[f'strength_{p}'].fillna(value=np.nan).dropna().values, 
                       exploded_merged_control[f'strength_{p}'].fillna(value=np.nan).dropna().values,
         alternative='greater'))
    
    
    mm_agg = pd.DataFrame(mm_loops.groupby(f'{lt}_loop_classes_{p}').size())/len(mm_loops)
    control_agg = pd.DataFrame(exploded_merged_control.groupby(f'{lt}_loop_classes_{p}').size())/len(exploded_merged_control)
    df = pd.merge(mm_agg, control_agg, on=f'{lt}_loop_classes_{p}', suffixes=('MM', control_nametag))

    df.plot.bar(color=pal[lt])
    plt.title(f"{len(mm_loops)} at MM, {len(exploded_merged_control)} at {control_nametag}")
    #plt.ylim(0, 1200)
    if writepath:
        plt.savefig(os.path.join(writepath, f"loop_numbers_{lt}_{p}_{tag}.svg"))
    plt.show()
    
    
    df.plot.pie(subplots=True, legend=False, fontsize=12, figsize=(12, 6))
    if writepath:
        plt.savefig(os.path.join(writepath, f"loop_pie_{lt}_{p}_{tag}.svg"))
    plt.show()
    
    pass
```

```python
def plot_hists_clean(exploded_merged, exploded_merged_control, p, 
                           lt='crosser', quant_sub=((2,3)), writepath=None, 
                           loop_score_types=('oe','able'), tag='mm'):
    score_map = {'oe':f'{lt}_loop_scores', 'able':'strength'}
    for loop_score in loop_score_types:
        for quant in quant_sub:
            mm_loops = exploded_merged[exploded_merged['quant'].isin(quant)]
            sns.histplot(mm_loops[f'{score_map[loop_score]}_{p}'].dropna(), bins=50, label=f'{p} MM {quant}',
                        stat='density', element='step', fill=False, color=two_colors[lt][0], linewidth=2)
            print(f"n={len(mm_loops[f'{score_map[loop_score]}_{p}'].dropna())}")
        sns.histplot(exploded_merged_control[f'{score_map[loop_score]}_{p}'].dropna(), bins=50, label=f'{p} {control_nametag}',
                stat='density', element='step', fill=False, color=two_colors[lt][1], linewidth=2)
        print(f"n={len(exploded_merged_control[f'{score_map[loop_score]}_{p}'].dropna())}")
        plt.xlim([-0.05, 0.5])
        plt.legend()
        if writepath is not None:
            plt.savefig(os.path.join(writepath, f"{loop_score}_hist_{lt}_{p}_{tag}.svg"))
        plt.show()


def plots_crosser_ss_clean(exploded_merged, exploded_merged_control, p, 
                           lt='crosser', quant_sub=((2,3)), writepath=None, 
                           loop_score_types=('oe','able'), tag='mm'):
    for loop_score in loop_score_types:
        tidy_plot = pd.DataFrame()
        
        score_map = {'oe':f'{lt}_loop_scores', 'able':'strength'}
        ylim_map = {'oe': (-0.1, 20), 'able':(-0.05, 0.5)}
        
        for quant in quant_sub:
            mm_loops = exploded_merged[exploded_merged['quant'].isin(quant)]
            tidy_mm = pd.DataFrame()
            tidy_mm['loop score'] = mm_loops[f'{score_map[loop_score]}_{p}'].dropna()
            tidy_mm['element class'] = f'MM {quant}'
            tidy_plot = pd.concat([tidy_plot, tidy_mm], axis=0)
        
        tidy_c = pd.DataFrame()
        tidy_c['loop score'] = exploded_merged_control[f'{score_map[loop_score]}_{p}'].dropna()
        tidy_c['element class'] = control_nametag
        tidy_plot = pd.concat([tidy_plot, tidy_c])
        means = tidy_plot.groupby("element class")["loop score"].mean()

        print(tidy_plot)
        f, ax = plt.subplots(figsize=(4,4)) 
        ax = sns.violinplot(tidy_plot, x='element class', y='loop score', hue='element class', legend="full",palette=pal[lt])
        
        handles, labels = ax.get_legend_handles_labels()
        new_labels = [f"{label} (mean = {means[label]:.3f})" for label in labels]
        
        ax.legend(handles, new_labels)
        plt.ylim(ylim_map[loop_score])

        if writepath:
            plt.savefig(os.path.join(writepath, f"{loop_score}_violin_{lt}_{p}_{tag}.svg"))
        plt.show()
        
        groups = [
            vals['loop score'].fillna(value=np.nan).dropna().values
            for _, vals in tidy_plot.groupby('element class')
        ]

        H, p_kw = kruskal(*groups)
        num_combo = len([*combinations(tidy_plot['element class'].unique(), 2)])
        for g1, g2 in combinations(tidy_plot['element class'].unique(), 2):
            x = tidy_plot.loc[
                tidy_plot['element class'] == g1,
                'loop score'
            ].fillna(value=np.nan).dropna()

            y = tidy_plot.loc[
                tidy_plot['element class'] == g2,
                'loop score'
            ].fillna(value=np.nan).dropna()

            U, pw = mannwhitneyu(x, y, alternative='two-sided')

            print(f"{g1} vs {g2}: U={U:.1f}, p={pw:.3e}")
    
    dfs = []

    for quant in quant_sub:
        mm_loops = exploded_merged[exploded_merged['quant'].isin(quant)]

        agg = (
            mm_loops
            .groupby(f'{lt}_loop_classes_{p}')
            .size()
            .div(len(mm_loops))
            .to_frame(name=f'MM {quant}')
        )

        dfs.append(agg)

    control_agg = (
        exploded_merged_control
        .groupby(f'{lt}_loop_classes_{p}')
        .size()
        .div(len(exploded_merged_control))
        .to_frame(name=control_nametag)
    )

    dfs.append(control_agg)

    df = pd.concat(dfs, axis=1).fillna(0)
    ax = df.plot.bar(figsize=(6,4),color=pal[lt])

    if writepath:
        plt.savefig(
            os.path.join(
                writepath,
                f"loop_numbers_{lt}_{p}_{tag}.svg"
            )
        )

    plt.show()
    
    pass
```

```python
def run_crosser_ss(test_df, control_df, loops_able, quant_subs=(2, 3), 
                   phases=['P1', 'P2', 'P3'], writepath=None, tag='mm', loop_score_types=('oe','able')):
    print(len(loops_able))
    for p in phases:
        for lt in ['ss', 'crosser']:
            if lt == 'ss':
                exploded = explode_ss(test_df, p=p)
                exploded_control = explode_ss(control_df, p=p)
            else:
                exploded = explode_df(test_df, p=p)
                exploded_control = explode_df(control_df, p=p)
            
            exploded_merged = pd.merge(exploded,loops_able,on=['chrom1','coord1','coord2'],
                                       how="inner", suffixes=('','_lc'))
            exploded_merged_control = pd.merge(exploded_control,loops_able,on=['chrom1','coord1','coord2'],
                                       how="inner", suffixes=('','_lc'))

            #plots_crosser_ss(exploded_merged, exploded_merged_control, p, lt=lt, writepath=writepath, tag=tag)
            plots_crosser_ss_clean(exploded_merged, exploded_merged_control, p, lt=lt, 
                                   writepath=writepath, tag=tag, quant_sub=quant_subs,
                                  loop_score_types=loop_score_types)
            plot_hists_clean(exploded_merged, exploded_merged_control, p, lt=lt, 
                                   writepath=writepath, tag=tag, quant_sub=quant_subs,
                                  loop_score_types=loop_score_types)
```

```python
def loop_strength_by_distance(merged_loops_sub, col1, col2=None, tstr='loop strengths by distance', sd=None):
    plt.figure()
    toplot=[]
    dist_range = [10000,20000,30000,40000,50000,60000,70000,80000,90000,100000]
    for i, dist in enumerate(dist_range):
        if i == len(dist_range)-1:
            merged_loops_dist = merged_loops_sub[(abs(merged_loops_sub['coord1']-merged_loops_sub['coord2'])>dist)]
        else:
            merged_loops_dist = merged_loops_sub[(abs(merged_loops_sub['coord1']-merged_loops_sub['coord2'])>dist)
                                                    & (abs(merged_loops_sub['coord1']-merged_loops_sub['coord2'])<dist_range[i+1])]
        if col2 is None:
            med = merged_loops_dist.loc[:,f"{col1}"]
        else:
                med = merged_loops_dist.loc[:,f"strength_{col1}"].divide(merged_loops_dist.loc[:,f"strength_{col2}"])
        toplot.append(med.median())
    plt.scatter(dist_range, toplot)
    plt.plot(dist_range, toplot)
    plt.title(tstr)
    plt.xlabel('Genomic Separation (bp)')
    plt.ylabel('Loop Strength')
```

```python
mm_loops_sub = mm_loops[mm_loops['crosser_loop_classes_P3'].isin(class_list)]
mm_loops_sub = mm_loops_sub[mm_loops_sub[['chrom1','coord']].duplicated(keep=False)]

mm_loops_sub['dist1'] = mm_loops_sub['coord'] - mm_loops_sub['coord1']
mm_loops_sub['dist2'] = mm_loops_sub['coord2'] - mm_loops_sub['coord']
mm_loops_sub['euc_dist'] = np.sqrt(mm_loops_sub['dist2']**2 + mm_loops_sub['dist1']**2)

idxmin = mm_loops_sub.groupby(['chrom1','coord'])['euc_dist'].idxmin()
mm_loops_closest = mm_loops_sub.loc[idxmin]

mm_loops_not_closest = mm_loops_sub[~mm_loops_sub.index.isin(idxmin)]
idxmin2 = mm_loops_not_closest.groupby(['chrom1','coord'])['euc_dist'].idxmin()
mm_loops_2close = mm_loops_not_closest.loc[idxmin2]

```
