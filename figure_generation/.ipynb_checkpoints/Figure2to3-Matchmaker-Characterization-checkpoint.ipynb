```python
def do_kmeans(df, nclus, rand=None):
    clus = cluster.KMeans(n_clusters=nclus, random_state=rand)
    return clus.fit_predict(df)

def plot_clusters(df, grp=None, writepath=None):
    if grp is None:
        sns.heatmap(df)
        plt.show()
    else:
        cm = dict(zip(np.unique(grp), sns.color_palette('hls', n_colors=len(np.unique(grp)))))
        df['g'] = grp
        df.sort_values(by='g', inplace=True)
        print(df)
        sns.clustermap(df.drop(columns='g', axis=1), row_colors=[cm[g] for g in df['g']], 
                       row_cluster=False, col_cluster=False, cmap='coolwarm', center=1)
        

        if writepath is not None:
            plt.savefig(f"{writepath}.svg", dpi=300, format='svg')
        else:
            plt.show()

    return df
```

```python
# loop scores by class for each phase 

def get_loops_by_row(row, loop_type, phase_num, class_list, how='sum'):
    
    inds = [i for i in range(len(row[f"{loop_type}_loop_classes_P{phase_num}"])) if row[f"{loop_type}_loop_classes_P{phase_num}"][i] in class_list]
    loops = [row[f"{loop_type}_loop_scores_P{phase_num}"][i] for i in inds if row[f"{loop_type}_loop_scores_P{phase_num}"][i] is not None]
    
    if how=='sum':
        return sum(loops) # add other agg methods later
    elif how=='mean':
        return np.mean(loops)
    
def loop_scores_by_class(df_in, loop_type, class_list, class_list_name, phase_range=range(1, 4), how='sum'):
    df = df_in.copy()
    for phase_num in phase_range:
        df[f'class_score_{class_list_name}_P{phase_num}'] = df.apply(lambda row: get_loops_by_row(row, loop_type, phase_num, class_list, how), axis=1)
    
    return df
```

```python
match_scores_classwise = loop_scores_by_class(match_scores, 'crosser', class_list=['CTCF-CTCF'], class_list_name='CTCF',
                                             how='mean')
match_scores_classwise = loop_scores_by_class(match_scores_classwise, 'crosser', class_list=['E-CTCF','P-CTCF'], class_list_name='CRE',
                                             how='mean')
```
