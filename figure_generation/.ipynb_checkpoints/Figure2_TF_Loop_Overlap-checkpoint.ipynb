```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import fisher_exact
import itertools

plt.rcParams.update({
    "text.usetex": False})
plt.rcParams['svg.fonttype'] = 'none'
plt.rc('pdf',fonttype = 42)
plt.rcParams["ps.useafm"] = True
```

```python
def fdr_bh(pvals):
    """
    Benjamini-Hochberg FDR correction.
    Returns array of adjusted p-values (same order as input).
    """
    pvals = np.array(pvals)
    n = len(pvals)

    # Sort p-values
    order = np.argsort(pvals)
    ranked_pvals = pvals[order]

    # Compute BH values
    bh = ranked_pvals * n / (np.arange(1, n+1))

    # Enforce monotonicity (important step)
    bh = np.minimum.accumulate(bh[::-1])[::-1]

    # Cap at 1
    bh = np.minimum(bh, 1.0)

    # Return to original order
    adjusted = np.empty(n)
    adjusted[order] = bh

    return adjusted
```

```python
def plot_tf_enrichment(df,
                       alpha=0.01,
                       pval=0.01,
                       only_enriched=False,
                       sort=True,
                       title="Significant TF Enrichment in Loop Anchors",
                       sd=None):
    """
    Parameters
    ----------
    df : pandas DataFrame
        Must contain columns:
        ["tf_overlap", "control_overlap",
         "tf_not_overlap", "control_not_overlap"]
        Index should be TF names.
    
    alpha : float
        FDR threshold (default 0.05)

    only_enriched : bool
        If True, show only enriched TFs (log2 OR > 0)

    sort : bool
        Sort by log2 odds ratio before plotting

    Returns
    -------
    df : dataframe with added columns:
        odds_ratio, log2_or, pval, fdr
    """

    df = df.copy()

    # ---- Fisher test ----
    odds_ratios = []
    pvals = []

    for _, row in df.iterrows():
        table = [[row["tf_overlap"], row["tf_not_overlap"]],
                 [row["control_overlap"], row["control_not_overlap"]]]
        
        or_val, p_val = fisher_exact(table, alternative="two-sided")
        
        odds_ratios.append(or_val)
        pvals.append(p_val)

    df["odds_ratio"] = odds_ratios
    df["log2_or"] = np.log2(df["odds_ratio"])
    df["pval"] = pvals

    df["fdr"] = fdr_bh(df["pval"])

    # ---- Filter significant ----
    sig_df = df[(df["fdr"] < alpha)&(df["pval"] < pval)].copy()
    
    if only_enriched:
        sig_df = sig_df[sig_df["log2_or"] > 0]

    if sig_df.empty:
        print("No significant TFs found.")
        return df

    if sort:
        sig_df = sig_df.sort_values("log2_or")

    # ---- Plot ----
    colors = ["#CC3311" if x > 0 else "#0077BB"
              for x in sig_df["log2_or"]]

    plt.figure()
    plt.barh(sig_df.index, sig_df["log2_or"], color=colors)
    plt.axvline(0)
    plt.xlabel("log2(Odds Ratio)")
    plt.title(title + f"\n(FDR < {alpha})")
    plt.tight_layout()
    
    if sd is not None:
        plt.savefig(sd)
    else:
        plt.show()

    return df
```

```python
# compute odds
df = pd.read_csv(os.path.join(basedir, input_file), sep="\t", header=None,
                 names=["TF", "tf_overlap", "control_overlap",
                        "tf_not_overlap", "control_not_overlap"])

df = df.set_index("TF")

# ---- Compute odds ratio ----
# Add small pseudocount to avoid division by zero
seudocount = 0.5
a = df["tf_overlap"] + pseudocount
b = df["tf_not_overlap"] + pseudocount
c = df["control_overlap"] + pseudocount
d = df["control_not_overlap"] + pseudocount

df["log2_or"] = np.log2((a * d) / (b * c))

# Sort
df = df.sort_values("log2_or")

# Plot
plt.figure()
plt.barh(df.index, df["log2_or"])
plt.axvline(0)
plt.xlabel("log2(Odds Ratio)")
plt.title("TF Enrichment in Loop Anchors")
plt.tight_layout()
plt.show()
```
